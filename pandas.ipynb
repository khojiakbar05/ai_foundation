```python
data_with_missing = {
    "cars": ["BMW", "Volvo", None, "Tesla", "Ferrari", "Ford"],
    "counts": [3, 5, None, 3, 2, None]
}

df.missing = pd.DataFrame(data_with_missing)

print(df.missing)
```

```python
df.missing.isnull().sum()
# bu booleanda chiqazib beradi
# .sum() nullarni hisoblab olib beradi
```

```python
df.missing.dropna(subset=['counts'])
# null bolgan qiymat qatorlarni tushurib qoldiradi
# (how="all") qatori bilan ochirib tashlaydi
# (axis=1) - qaysi columnda malumot yoq bolsa o'shani ochirib tashlaydi
```

```python
# df.missing.fillna('unknown')
# data yoq bolgan qaotlarga qiymat qoshib beradi

df.missing.fillna({'cars': "unknown", "counts": 0})
# carsga unknown || countsga 0 qoyib beradi
```
