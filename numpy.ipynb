```python
import numpy as np

arr = np.array([1,3, 5, 7])
print('numpy_array: ', arr)

arr.shape
```

```python
# shape - qator va ustunlar sonini aytadi

arr2 = np.array([[1,3, 5, 7] , [4, 6, 8, 10]])
print('numpy_array: ', arr2)

arr2.shape
```

```python
# dtype () - Bu massivdagi elementlarning ma’lumot turini bildiradi.

a = np.array([1, 2, 3])
print(a.dtype)
```

```python
# reshape() - massivning ma’lumotlarini o’zgartirmasdan faqat shaklini o’zgartiradi

a = np.array([1, 2, 3, 4, 5, 6])
b = a.reshape(2, 3)
print("shaped_array: ", b)
```

```python
# astype() - massiv elementlarining turini o’zgartiradi va yangi massiv qaytaradi.

a = np.array([1, 2, 3])
b = a.astype(float)
print("float: ", b)
```

```python
# np.zeros(), np.ones() => dummy
# ikkalasi ham funksiya bo’lib, berilgan shakldagi massivni mos ravishda nollar yoki birlar bilan to’ldirib yaratadi.
# Ular ko’pincha “dummy data” (vaqtincha, boshlang’ich massiv) sifatida ishlatiladi.

z = np.zeros((2, 3))
o = np.ones((2, 3))

print("zeros: ", z)
print("ones: ", o)
```

```python
# ndim - massivning o’lchamlari sonini bildiradi

a = np.array([1, 2])
print("ndim: ", a)
```

```python
# arange() - berilgan oraliqda ketma-ket sonlar massivini yaratadi.

a = np.arange(1, 6)

print("arange: ", a)
```

```python
# sum(), mean(), max(), min()

# sum() → yig’indi (1+2+3+4 = 10)
# mean() → o’rtacha qiymat (10 / 4 = 2.5). Natija doim float bo’ladi
# max() → eng katta element
# min() → eng kichik element

a = np.array([1, 2, 3, 4])

print("sum: ", a.sum())
print("mean: ", a.mean())
print("max: ", a.max())
print("min: ", a.min())
```

```python
# flatten().    1D arrayni 2D arrayga aylantirib beradi (rasmlarda ishlatilinadi)

a = np.array([[1, 2], [3, 4]])

print("flatten: ", a.flatten())
```

```python
# T (Transpose) - massivni transpozitsiya qiladi: qatorlar ustunga, ustunlar qatorga aylanadi

a = np.array([[1, 2], [3, 4]])

print("transpose: ", a.T)
```

```python
# dot() @.   matritsalarni kopaytirish uchun matematik amal

a = np.array([[1, 2], [3, 4]])
b = np.array([[2, 0], [1, 2]])

print("dot/multiplication: ", a.dot(b))
```

```python
# random.rand() render paytda turli xil raqamlarni render qilib beradi

a = np.random.rand(2, 3) # => 0.0 and 1.0 / 0.0 - included  / 1.0 excluded
print("random_rand: ", a)
```
